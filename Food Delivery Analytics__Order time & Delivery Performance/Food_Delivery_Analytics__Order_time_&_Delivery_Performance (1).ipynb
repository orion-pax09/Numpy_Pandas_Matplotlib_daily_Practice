```python
df = pd.read_csv("Food delivery analytics order time & delivery performance.csv")
print(df)
```

```python
Df = df.copy()
print(Df)
```

```python
Df["delivered_time"] = pd.to_datetime(Df["delivered_time"])
print(Df["delivered_time"])
print(Df.dtypes)
```

```python
print(Df)
```

```python
Df["Day_Name"] = Df["delivered_time"].dt.day_name()  
print(Df)
```

```python
Df["order_time"] = pd.to_datetime(Df["order_time"])
print(Df)
```

```python
Df.loc[
    Df["status"] == "Cancelled",
    "delivered_time"
] = pd.NaT
Df["delivered_time"] = pd.to_datetime(Df["delivered_time"])
Df["order_time"] = pd.to_datetime(Df["order_time"])
Df["delivery_minutes"] = (Df["delivered_time"] - Df["order_time"]).dt.total_seconds() / 60  
print(Df)
```

```python
condition = [Df["delivery_minutes"] < 30 , (Df["delivery_minutes"]>=30)&(Df["delivery_minutes"]<=40) , Df["delivery_minutes"] > 40]
choice = ["Fast" , "Normal" , "Slow"]
delivery_category = np.select(condlist = condition , choicelist = choice , default = "Cancelled")
Df["delivery_category"] = delivery_category
print(Df)
```

```python
flt = Df.groupby("city").agg(
    total_order = ("city" , "count"),
    average_delivery = ("delivery_minutes" , "mean"))
print(flt)
```

```python
completed_orders = Df[Df["status"]=="Completed"].groupby(by="city").size()
flt["completed_orders"] = completed_orders
print(flt)
```

```python
cancelled_orders = Df[Df["status"]=="Cancelled"].groupby(by="city").size()
flt["Cancelled_orders"] = cancelled_orders
print(flt)
```

```python
count_orders = Df["Hour"].value_counts().sort_index()
print(count_orders)
```

```python
busiest_hour = count_orders.idxmax()
print(busiest_hour)
```

```python
plt.bar(count_orders.index , count_orders.values , color = "black")
plt.grid(linewidth = 1)
plt.xlabel("Hours (t) axis")
plt.ylabel("No. of orders")
plt.title("Orders per hours")
plt.show()
```
