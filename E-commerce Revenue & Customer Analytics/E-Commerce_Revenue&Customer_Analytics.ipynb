```python
merge = pd.merge(left = order_copy , right = customer , on ="customer_id")
print(merge)
```

```python
quantity= merge["quantity"].to_numpy()
price = merge["price"].to_numpy()
print(quantity)
print(price)
```

```python
revenue = price * quantity
print(revenue)
```

```python
merge["total_price"] = revenue
print(merge)
```

```python
membership_summary = merge.groupby(by="city").agg(
    total_orders = ("order_id" , "count"),
    total_revenue = ("total_price" , "sum"),
    average_revenue = ("total_price","mean")
)
print(membership_summary)
```

```python
pivot = pd.pivot_table(
    merge,
    index = "city",
    columns = "category",
    values = "total_price",
    aggfunc = "sum"
)
print(pivot)
```

```python
customer_summary = merge.groupby(by="customer_id").agg(
    total_spending = ("total_price" , "sum"),
    city = ("city" , "first"),
    membership = ("membership" , "first"),
)
print(customer_summary)
```

```python
highest_spending = customer_summary["total_spending"].idxmax()
print(highest_spending)
```

```python
print(customer_summary.loc[[highest_spending]])
```

```python
category_revenue = merge.groupby(by="category")["total_price"].sum()
print(category_revenue)
```

```python
plt.title("Total revenue by category")
plt.bar(category_revenue.index ,category_revenue.values , color = "purple")
plt.xticks(rotation = 20)
plt.xlabel("Category items")
plt.ylabel("total prices")
plt.tight_layout()
plt.show()
```
