```python
merge = pd.merge(left = order_copy , right = customer , on ="customer_id")
print(merge)
```

```python
quantity= merge["quantity"].to_numpy()
price = merge["price"].to_numpy()
print(quantity)
print(price)
```

```python
revenue = price * quantity
print(revenue)
```

```python
del merge["total price"]
merge["total_price"] = revenue
print(merge)
```

```python
membership_summary = merge.groupby(by="category").agg(
    total_orders = ["category" , "count"],
    total_revenue = ["revenue" , "sum"],
    average_revenue = ["revenue","mean"])
print(membership_summary)
```
